## Trực quan hóa kết quả (Ma trận nhầm lẫn & ROC/PR)
Chạy `python evaluation/metrics/calculate_metrics.py` trước khi xem các hình ảnh này.

```python
metrics_dir = Path('../evaluation/metrics')
modes = ['mode_a', 'mode_b']

for mode in modes:
    print(f'
--- Trực quan hóa cho {mode.upper()} ---')
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    cm_path = metrics_dir / f'confusion_matrix_{mode}.png'
    if cm_path.exists():
        axes[0].imshow(mpimg.imread(cm_path))
        axes[0].axis('off')
        axes[0].set_title('Ma trận nhầm lẫn')
    
    roc_path = metrics_dir / f'roc_curve_{mode}.png'
    if roc_path.exists():
        axes[1].imshow(mpimg.imread(roc_path))
        axes[1].axis('off')
        axes[1].set_title('Biểu đồ ROC')
        
    pr_path = metrics_dir / f'pr_curve_{mode}.png'
    if pr_path.exists():
        axes[2].imshow(mpimg.imread(pr_path))
        axes[2].axis('off')
        axes[2].set_title('Biểu đồ Precision-Recall')
        
    plt.show()
```
